```python
# importing libraries
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("final_weather_traffic_dataset.csv")
print(df.shape)
print(df.info())
print(df.head())
```

```python
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'])
df.info() 
```

```python
# checking missing values
df.isnull().sum().sort_values(ascending=False)

```

```python
df = df.dropna(subset=[
    'fare_amount',
    'pickup_datetime',
    'pickup_latitude',
    'pickup_longitude',
    'dropoff_latitude',
    'dropoff_longitude',
    'distance_km'
])
```

```python
df['fare_amount'].fillna(df['fare_amount'].median(), inplace=False)
df['distance_km'].fillna(df['distance_km'].median(), inplace=False)
```

```python
df['temperature'] = df['temperature'].fillna(df['temperature'].median())
df['precipitation'] = df['precipitation'].fillna(0)
df['wind_speed'] = df['wind_speed'].fillna(df['wind_speed'].median())
```

```python
df['duration_min'] = df['duration_min'].fillna(df['duration_min'].median())
df['traffic_duration_min'] = df['traffic_duration_min'].fillna(df['traffic_duration_min'].median())
df['traffic_delay'] = df['traffic_delay'].fillna(0)
df['avg_speed'] = df['avg_speed'].fillna(df['avg_speed'].median())
```

```python
df = df.dropna(subset=['pickup_datetime'])
```

```python
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'])
```

```python
numeric_cols = [
    'fare_amount',
    'pickup_longitude',
    'pickup_latitude',
    'dropoff_latitude',
    'dropoff_longitude',
    'passenger_count',
    'temperature',
    'precipitation',
    'wind_speed',
    'distance_km',
    'duration_min',
    'traffic_duration_min',
    'traffic_delay',
    'avg_speed'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df['is_peak'] = df['is_peak'].astype(bool)
```

```python
df.info()
```

```python
df['distance_km'].describe()
```

```python
df = df[df['distance_km'] > 0]
df = df[df['distance_km'] < 100]   # city ride assumption
```

```python
df = df[df['fare_amount'] > 0]
df = df[df['fare_amount'] < 500]   # remove extreme
```

```python
df = df[(df['pickup_latitude'].between(-90,90)) &
        (df['dropoff_latitude'].between(-90,90))]

df = df[(df['pickup_longitude'].between(-180,180)) &
        (df['dropoff_longitude'].between(-180,180))]
```

```python
df['avg_speed'].describe()
```

```python
df = df[df['avg_speed'] < 150]
```

```python
Q1 = df['fare_amount'].quantile(0.25)
Q3 = df['fare_amount'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['fare_amount'] >= Q1 - 1.5*IQR) &
        (df['fare_amount'] <= Q3 + 1.5*IQR)]
```

```python
Q1 = df['distance_km'].quantile(0.25)
Q3 = df['distance_km'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['distance_km'] >= Q1 - 1.5*IQR) &
        (df['distance_km'] <= Q3 + 1.5*IQR)]
```

```python
Q1 = df['duration_min'].quantile(0.25)
Q3 = df['duration_min'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['duration_min'] >= Q1 - 1.5*IQR) &
        (df['duration_min'] <= Q3 + 1.5*IQR)]
```

```python
df['pickup_latitude'] = df['pickup_latitude'].round(5)
df['pickup_longitude'] = df['pickup_longitude'].round(5)
df['dropoff_latitude'] = df['dropoff_latitude'].round(5)
df['dropoff_longitude'] = df['dropoff_longitude'].round(5)
```

```python
df['distance_km'] = df['distance_km'].round(2)
```

```python
df['fare_amount'] = df['fare_amount'].round(2)
```

```python
df['avg_speed'] = df['avg_speed'].round(2)
```

```python
df['temperature'] = df['temperature'].round(1)
```

```python
df.to_csv("cleaned_weather_traffic_dataset.csv", index=False)
```
